# rule_swap_analysis_v2 演示样例

本 notebook 全面演示 `rule_swap_analysis_v2` 的各项功能，覆盖以下场景：

1. **基础用法** - reference_data 自动计算分箱表
2. **传入现成分箱表** - 直接使用预计算的分箱表
3. **多评分场景** - 多个评分联合分析
4. **多标签场景** - overdue + dpds 多期逾期分析
5. **OUT-OUT 拒绝规则分析** - rules_base 基准拒绝规则
6. **IN-OUT 置出规则分析** - 规则置出（误拒）分析
7. **OUT-IN 置入规则分析** - 规则置入（捞回）分析
8. **完整四象限 Swap 分析** - rules_base + rules_out + rules_in 联合
9. **sequential vs independent 模式对比**
10. **金额口径分析** - amount 字段加权
11. **自定义分箱参数** - bin_method / max_n_bins / min_bin_size
12. **可视化与结果解读**

```python
import os, sys
sys.path.append('../')
```

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

from hscredit.core.rules import Rule
from hscredit.report.rule_analysis import rule_swap_analysis_v2
from hscredit.report.feature_analyzer import feature_bin_stats

# 设置中文显示
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 200)
pd.set_option('display.float_format', '{:.4f}'.format)

print('hscredit version:', __import__('hscredit').__version__)
```

## 0. 生成模拟数据

生成一个有表现样本（reference_data）和待分析样本（data），包含评分、特征、金额字段。

```python
np.random.seed(42)
n_ref = 5000   # 历史有表现样本数
n_data = 2000  # 待分析样本数

# ---- 历史有表现样本（用于计算分箱表）----
ref_score = np.random.normal(620, 80, n_ref)
ref_age = np.random.randint(22, 55, n_ref)
ref_income = np.random.randint(3000, 30000, n_ref)
ref_amount = np.random.lognormal(10.5, 0.8, n_ref)  # 贷款金额

# 评分越高，target=0（正常）概率越高（模拟真实风控逻辑）
ref_target_prob = 1 / (1 + np.exp((620 - ref_score) / 100))
ref_target = (np.random.random(n_ref) < ref_target_prob).astype(int)

reference_data = pd.DataFrame({
    'score_a': ref_score,
    'score_b': ref_score + np.random.normal(10, 20, n_ref),
    'age': ref_age,
    'income': ref_income,
    'amount': ref_amount,
    'target': ref_target,
})
reference_data['score_a'] = reference_data['score_a'].clip(300, 850)
reference_data['score_b'] = reference_data['score_b'].clip(300, 850)

print(f'有表现样本: {len(reference_data)}, 坏样本率: {reference_data["target"].mean():.4f}')
print(f'金额范围: [{reference_data["amount"].min():.0f}, {reference_data["amount"].max():.0f}]')
reference_data.head(3)
```

```python
# ---- 待分析样本（模拟当前进件）----
data_score = np.random.normal(610, 85, n_data)
data_age = np.random.randint(22, 55, n_data)
data_income = np.random.randint(3000, 30000, n_data)
data_amount = np.random.lognormal(10.5, 0.8, n_data)

data = pd.DataFrame({
    'score_a': data_score,
    'score_b': data_score + np.random.normal(10, 20, n_data),
    'age': data_age,
    'income': data_income,
    'amount': data_amount,
})
data['score_a'] = data['score_a'].clip(300, 850)
data['score_b'] = data['score_b'].clip(300, 850)

print(f'待分析样本: {len(data)}')
data.head(3)
```

## 场景 1：基础用法 - reference_data 自动计算分箱表

只需传入 `reference_data` 和 `target`，自动从数据计算评分分箱表。

```python
# 定义一条 OUT-OUT 拒绝规则：评分 < 580 则拒绝
rule_base = Rule('score_a < 580', name='评分偏低拒绝')

result = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base],
    reference_data=reference_data,
    target='target',
)

print('=== swap_pipeline (四象限汇总) ===')
display(result['swap_pipeline'])
print()
print('=== swap_result (分步骤通过率变化) ===')
display(result['swap_result'])
```

**结果解读**：
- `全量样本` 行显示基准坏样本率（分箱表预测）
- `OUT-OUT规则` 行显示评分 < 580 的拒绝样本的风险特征（LIFT > 1 说明高风险）
- `OUT-OUT合计` 行是所有拒绝规则的联合命中
- `剩余样本` 行是通过拒绝规则后的剩余样本（通过率 = 1 - 拒绝率）

## 场景 2：传入现成分箱表

使用预计算的分箱表，跳过 reference_data 计算过程，适合分箱表已由其他工具生成或缓存的场景。

```python
# 先用 feature_bin_stats 计算分箱表
bin_table = feature_bin_stats(
    reference_data,
    'score_a',
    target='target',
    max_n_bins=8,
    method='quantile',
)
print('预计算分箱表结构:')
display(bin_table)
print('分箱表列名:', bin_table.columns.tolist())
```

```python
# 传入现成分箱表
result2 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base],
    bin_table=bin_table,   # 直接传入现成分箱表
)
display(result2['swap_pipeline'])
```

传入现成分箱表与传入 reference_data 的结果基本一致（微小差异来自分箱边界略有不同）。

## 场景 3：多评分场景

传入 Dict[str, str] 格式的多评分，多个评分加权计算预测坏概率。

```python
# 两套评分联合分析
multi_score = {'model_a': 'score_a', 'model_b': 'score_b'}

rule_base2 = Rule('(score_a < 580) & (score_b < 560)', name='评分组合拒绝')

result3 = rule_swap_analysis_v2(
    data=data,
    score=multi_score,
    rules_base=[rule_base2],
    reference_data=reference_data,
    target='target',
)
display(result3['swap_pipeline'])
```

多评分场景下，`bin_table` 可传入 Dict[str, pd.DataFrame]，每个评分独立计算：

```python
# 多评分 + 权重
bin_table_multi = {
    'model_a': feature_bin_stats(reference_data, 'score_a', 'target', max_n_bins=8),
    'model_b': feature_bin_stats(reference_data, 'score_b', 'target', max_n_bins=8),
}

result3b = rule_swap_analysis_v2(
    data=data,
    score=multi_score,
    rules_base=[rule_base2],
    bin_table=bin_table_multi,
    score_weights={'model_a': 0.6, 'model_b': 0.4},  # 模型A权重更高
)
display(result3b['swap_pipeline'])
```

## 场景 4：多标签场景（overdue + dpds）

分析不同账龄（MOB）和逾期天数（dpd）组合下的置入置出效果。

```python
# 生成多期逾期数据
n_ref2 = 3000
ref_score2 = np.random.normal(620, 80, n_ref2)
ref_target2 = np.random.binomial(1, 0.15, n_ref2)  # 基准逾期率 15%

# 模拟 MOB1、MOB3、MOB6 三期的逾期标记
ref_mob1 = np.minimum((ref_target2 * np.random.randint(1, 91, n_ref2)), 90).astype(int)
ref_mob3 = np.minimum((ref_target2 * np.random.randint(1, 181, n_ref2)), 180).astype(int)
ref_mob6 = np.minimum((ref_target2 * np.random.randint(1, 271, n_ref2)), 270).astype(int)

ref_data_multi = pd.DataFrame({
    'score_a': ref_score2,
    'MOB1': ref_mob1,
    'MOB3': ref_mob3,
    'MOB6': ref_mob6,
})
ref_data_multi['score_a'] = ref_data_multi['score_a'].clip(300, 850)

print('多标签样本坏样本率:')
for col in ['MOB1', 'MOB3', 'MOB6']:
    print(f'  {col} 坏样本率: {(ref_data_multi[col] > 0).mean():.4f}')
```

```python
result4 = rule_swap_analysis_v2(
    data=data.head(500),
    score='score_a',
    rules_in=[Rule('score_a >= 680', name='优质客户置入')],
    reference_data=ref_data_multi,
    overdue=['MOB1', 'MOB3', 'MOB6'],
    dpds=[0, 7, 30, 60],  # dpd > 0, dpd > 7, dpd > 30, dpd > 60
)
print('=== swap_summary (多标签) ===')
display(result4['swap_pipeline'])
```

## 场景 5：OUT-OUT 拒绝规则分析（rules_base）

分析多条基准拒绝规则各自的拒绝效果和组合效果。

```python
# 定义多条拒绝规则
rule_age = Rule('age < 25', name='年龄过小拒绝')
rule_income = Rule('income < 5000', name='收入过低拒绝')
rule_score = Rule('score_a < 580', name='评分偏低拒绝')

result5 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_age, rule_income, rule_score],
    reference_data=reference_data,
    target='target',
    rule_analysis_mode='independent',  # 每条规则独立应用
)
display(result5['swap_pipeline'])
```

```python
# sequential 模式：规则漏斗叠加效果
result5b = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_age, rule_income, rule_score],
    reference_data=reference_data,
    target='target',
    rule_analysis_mode='sequential',  # 漏斗模式
)
print('=== Sequential 模式（规则漏斗叠加）===')
display(result5b['swap_pipeline'])
```

**对比**：
- `independent` 模式：每条规则独立应用到全量样本，`OUT-OUT合计` 是各规则联合命中（去重）
- `sequential` 模式：规则按顺序叠加，后面的规则在前面的规则拒绝后的剩余样本上分析，体现漏斗效果

## 场景 6：IN-OUT 置出规则分析（rules_out）

IN-OUT：通过了 rules_base 但被 rules_out 拒绝的样本（误拒样本）。
用于分析现有规则中是否存在误拒优质客户的情况。

```python
rule_base6 = Rule('score_a < 600', name='评分基准拒绝')
rule_out6 = Rule('(score_a >= 580) & (income >= 8000)', name='误拒规则(收入达标可捞)')

result6 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base6],
    rules_out=[rule_out6],
    reference_data=reference_data,
    target='target',
)
display(result6['swap_pipeline'])
```

**结果解读**：
- `IN-OUT置出` 行显示被误拒的样本数（评分 580-600 且收入 ≥ 8000 的客户）
- 观察 LIFT 值，若 < 1 说明误拒样本实际是好客户（误拒成本高）
- `IN-IN通过` 行显示通过 rules_base 且通过 rules_out 的剩余好客户

## 场景 7：OUT-IN 置入规则分析（rules_in）

OUT-IN：被 rules_base 拒绝，但被 rules_in 捞回的样本。
这部分样本从拒绝侧被拉回通过侧，风险需要上浮评估。

```python
rule_base7 = Rule('score_a < 600', name='评分基准拒绝')
rule_in7 = Rule('(score_a >= 570) & (income >= 15000)', name='高收入客户置入')

result7 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base7],
    rules_in=[rule_in7],
    reference_data=reference_data,
    target='target',
    out_in_uplift=2.0,  # OUT-IN 风险上浮 2 倍
)
display(result7['swap_pipeline'])
```

**结果解读**：
- `OUT-IN置入` 行显示从拒绝侧捞回的样本，其坏样本率已应用 `out_in_uplift`（默认 2.0）上浮
- `ALL-IN置换` 行 = IN-IN通过 + OUT-IN置入，表示全部应该通过的客户
- 通过率上涨 = OUT-IN 带来的额外通过率

```python
# 查看 swap_result（置换效果对比）
print('=== swap_result ===')
display(result7['swap_result'])
```

## 场景 8：完整四象限 Swap 分析

同时使用 rules_base + rules_out + rules_in，展示完整的 IN-IN / IN-OUT / OUT-IN / OUT-OUT 四象限分析。

```python
rule_base8 = Rule('score_a < 600', name='评分基准拒绝')
rule_out8 = Rule('(score_a >= 580) & (income >= 8000)', name='收入误拒捞回')
rule_in8 = Rule('(score_a >= 560) & (income >= 12000)', name='优质客户置入')

result8 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base8],
    rules_out=[rule_out8],
    rules_in=[rule_in8],
    reference_data=reference_data,
    target='target',
    out_in_uplift=2.0,
)
print('=== 四象限汇总 (swap_summary) ===')
display(result8['swap_pipeline'])
```

```python
print('=== 分步骤通过率变化 (swap_pipeline) ===')
display(result8['swap_pipeline'])
```

```python
print('=== 置换效果对比 (swap_result) ===')
display(result8['swap_result'])
```

**四象限解读**：

| 象限 | 定义 | 说明 |
|------|------|------|
| **OUT-OUT** | FAIL rules_base | 被基准拒绝规则拒绝 |
| **IN-OUT** | PASS rules_base, FAIL rules_out | 通过基准但被置出规则拒绝（误拒）|
| **IN-IN** | PASS rules_base, PASS rules_out | 最终通过的好客户 |
| **OUT-IN** | FAIL rules_base, PASS rules_in | 被基准拒绝但被置入规则捞回 |

**业务含义**：
- OUT-OUT 拒绝率高 → 规则偏严格
- IN-OUT 坏样本率低 → 误拒了优质客户，有优化空间
- OUT-IN LIFT 高 → 捞回的样本风险较高，上浮合理

## 场景 9：金额口径分析（amount 字段）

传入 `amount` 字段后，报告同时包含订单口径（按样本数）和金额口径（按贷款金额加权）的分析结果。

```python
rule_base9 = Rule('score_a < 600', name='评分偏低拒绝')

result9 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base9],
    reference_data=reference_data,
    target='target',
    amount='amount',  # 启用金额口径
)
display(result9['swap_pipeline'])
```

**金额口径解读**：
- 金额口径的 `坏样本率` 基于每笔贷款的逾期金额加权计算
- 高金额样本对坏样本率的贡献更大
- 对比两个口径的 `坏样本率` 和 `LIFT值`，可以发现样本口径和金额口径的风险差异

```python
# 查看分箱表结构（验证金额口径数据）
# 手动计算带金额的分箱表
bin_table_amount = feature_bin_stats(
    reference_data,
    'score_a',
    target='target',
    amount='amount',
    max_n_bins=8,
    method='quantile',
)
print('金额口径分箱表结构 (MultiIndex rows):')
display(bin_table_amount)
```

## 场景 10：自定义分箱参数

通过 `bin_params` 传入自定义分箱约束，如单调性、预定义切分点等。

```python
result10 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base],
    reference_data=reference_data,
    target='target',
    bin_method='quantile',     # 分箱方法：quantile / tree / chimerge
    max_n_bins=6,              # 最大分箱数
    min_bin_size=0.08,        # 每箱最小样本占比 8%
    bin_params={'monotonic': 'auto_asc_desc'},  # 强制单调性约束
)
display(result10['swap_pipeline'])
```

## 场景 11：sample_survival_rate 与 reverse_order

分析生产策略通过率不为 100% 的场景（如策略通过率 70%）。

```python
result11 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base],
    rules_in=[Rule('score_a >= 650', name='优质客户置入')],
    reference_data=reference_data,
    target='target',
    sample_survival_rate=0.7,  # 生产策略只通过 70% 的申请件
)
print('=== swap_summary (生产通过率 70%) ===')
display(result11['swap_pipeline'])
```

```python
print('=== swap_pipeline (逆序，从置入效果开始) ===')
result11b = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base],
    rules_in=[Rule('score_a >= 650', name='优质客户置入')],
    reference_data=reference_data,
    target='target',
    sample_survival_rate=0.7,
    reverse_order=True,  # 逆序：从置入效果开始展示
)
display(result11b['swap_pipeline'])
```

## 场景 12：OUT-IN 额度填充（out_in_amount_fill / out_in_amount_col）

对于 OUT-IN 捞回的样本，若数据中没有金额字段，可通过定值填充或字段映射。

```python
result12 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[Rule('score_a < 600', name='评分偏低拒绝')],
    rules_in=[Rule('(score_a >= 560) & (income >= 12000)', name='高收入置入')],
    reference_data=reference_data,
    target='target',
    out_in_amount_fill=10000.0,  # OUT-IN 样本固定填充 1 万元额度
    out_in_amount_col='amount', # 原始金额字段名（用于区分订单/金额口径）
)
print('=== OUT-IN 额度填充结果 ===')
display(result12['swap_pipeline'][result12['swap_pipeline']['规则分类'].isin(['OUT-IN置入', 'OUT-IN置入规则', 'ALL-IN置换'])])
```

## 场景 13：规则链式分析（多规则 IN-OUT + OUT-IN）

展示多规则组合的完整分析，包含 IN-OUT 误拒检测和 OUT-IN 捞回效果。

```python
rule_base13 = Rule('score_a < 600', name='评分基准')
rule_out13a = Rule('(score_a >= 580) & (income >= 10000)', name='优质误拒(高收入)')
rule_out13b = Rule('(score_a >= 570) & (score_a >= 620)', name='优质误拒(高评分)')
rule_in13a = Rule('(score_a >= 560) & (income >= 8000)', name='Vintage优质客户')
rule_in13b = Rule('income >= 15000', name='收入达标置入')

result13 = rule_swap_analysis_v2(
    data=data,
    score='score_a',
    rules_base=[rule_base13],
    rules_out=[rule_out13a, rule_out13b],
    rules_in=[rule_in13a, rule_in13b],
    reference_data=reference_data,
    target='target',
    rule_analysis_mode='independent',
    out_in_uplift=2.0,
)
print('=== 完整 Swap 分析 (多规则链式) ===')
display(result13['swap_pipeline'])
```

```python
print('=== 置换效果汇总 ===')
display(result13['swap_result'])
```

## 场景 14：可视化结果解读

提取关键指标，制作可视化图表辅助策略决策。

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 提取规则分析结果（v2: swap_pipeline 包含四象限汇总）
pipeline = result8['swap_pipeline']

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# 图1：各规则样本数与坏样本率
# v2 swap_pipeline 列：分析流程 / 规则集 / 指标名称 / 样本总数 / 坏样本率
rule_rows = pipeline[pipeline['规则分类'].isin(['OUT-OUT拒绝样本', 'IN-OUT置出样本', 'OUT-IN置入样本'])]
if not rule_rows.empty:
    ax1 = axes[0]
    x = range(len(rule_rows))
    labels = rule_rows['规则集'].tolist()
    bars = ax1.bar(x, rule_rows['样本总数'], color='steelblue', alpha=0.7)
    ax1.set_xticks(x)
    ax1.set_xticklabels(labels, rotation=30, ha='right', fontsize=8)
    ax1.set_ylabel('样本总数')
    ax1.set_title('各规则命中样本数')
    for bar, rate in zip(bars, rule_rows['坏样本率']):
        ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
                 f'{rate:.2%}', ha='center', va='bottom', fontsize=8)

# 图2：LIFT 值对比
if not rule_rows.empty:
    ax2 = axes[1]
    lift_vals = rule_rows['LIFT值'].fillna(0).tolist()
    colors = ['green' if l < 1 else 'red' for l in lift_vals]
    ax2.bar(x, lift_vals, color=colors, alpha=0.7)
    ax2.axhline(y=1, color='black', linestyle='--', linewidth=1)
    ax2.set_xticks(x)
    ax2.set_xticklabels(labels, rotation=30, ha='right', fontsize=8)
    ax2.set_ylabel('LIFT值')
    ax2.set_title('各规则 LIFT 值 (红线=1.0)')

# 图3：通过率变化流程
if not pipeline.empty:
    ax3 = axes[2]
    ax3.plot(range(len(pipeline)), pipeline['通过率(绝对值)'].fillna(0),
             marker='o', color='steelblue', linewidth=2)
    ax3.set_xticks(range(len(pipeline)))
    ax3.set_xticklabels(pipeline['规则分类'].tolist(), rotation=45, ha='right', fontsize=8)
    ax3.set_ylabel('通过率 (%)')
    ax3.set_title('策略调整后通过率变化')
    ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 场景 15：与 rule_swap_analysis v1 对比

`rule_swap_analysis_v2` 是 v1 的增强版，主要区别：

| 功能 | v1 | v2 |
|------|-----|-----|
| 分箱表来源 | reference_data 或 bin_table | 新增 reference_data 自动计算 |
| 多评分 | 支持 | 支持 |
| 多标签 | 不支持 | 支持（overdue+dpds）|
| 金额口径 | 不支持 | 支持（amount 参数）|
| sequential 模式 | 不支持 | 支持（规则漏斗叠加）|
| OUT-IN 额度填充 | 不支持 | 支持 |
| 自定义分箱 | 有限 | bin_method/max_n_bins/min_bin_size |
| 结果结构 | 两张表 | 两张表（pipeline/result）|

v1 与 v2 的主要 API 差异：

- **参数名**：v1 第一个参数为 `df`（positional），v2 为 `data`（keyword）
- **rules_in**：v1 需要 `rules_in` 作为必填参数，v2 中 `rules_in` 有默认值 `None`
- **新增能力**：v2 新增 `reference_data` 自动计算分箱、金额口径、sequential 模式等


## 总结

`rule_swap_analysis_v2` 提供了完整的规则置入置出（Swap）分析能力：

1. **四象限框架**：IN-IN / IN-OUT / OUT-IN / OUT-OUT 全面覆盖
2. **灵活输入**：支持 reference_data 自动计算分箱表或传入现成分箱表
3. **多维度分析**：多评分、多标签、金额口径、sequential/independent 模式
4. **完整报告**：swap_summary（四象限汇总）、swap_pipeline（通过率变化）、swap_result（置换效果）
5. **可解释指标**：LIFT、坏账改善、风险拒绝比、准确率、精确率、召回率等
